### **Execução da célula "#while2"**<br>
<p> A var "nr" recebeu o valor "1", onde ficou armazenada o respectivo dado.<br>
Após isso, criamos um loop com o While, onde indicamos que o "nr" deve sempre ser menor que "100".<br>
Com isso, damos a ordem de printar a var "nr".<br> Ele vai retornar o valor armazenado inicialmente na var "nr" e, após, somará mais 1, indicado na linha 8 "nr += 1"<br>
Isso tudo ocorrerá até que ele some as variáveis e o resultado seja igual a "5". Quando chegar ao "5", o loop será interrompido por causa da função "break".
</p>

```python
#while2
nr = 1
while nr < 100:
    print(nr)
    if nr == 5:
        break

    nr += 1
```

### **While dentro de While**
<p>Temos abaixo uma célula que recebe uma variável <code style="color: salmon;">linha</code>, que armazena o valor "0".<br>
Quando iniciamos o while, informamos a condição <code>while linha < 3:</code>, que diz "Enquanto a linha for menor q 3". A partir disso, geramos mais uma variável dentro do while, chamada <code style="color: salmon;">coluna</code> que recebe o valor "0".
Após à nova var, geramos outro loop, onde será <code><span style="color: #C678DD;">while</span> <span style="color: salmon;">coluna</span> <span style="color: cyan;"><</span> <code>3</code>:</code>
Ela vai printar a var linha e a var coluna. Com o While, ela trará o valor <code><span style="color: salmon;">linha</span> <span style="color: cyan;">+=</span> 1</code>

</P>

```python
linha = 0
while linha < 3:
    coluna = 0
    while coluna < 3:
        print("linha: ", linha, " - coluna: ", coluna)
        coluna += 1
    linha += 1
```

```python
nrInicial = 1

nrFinal = int(input("Digite um número maior que 1: "))
while nrInicial <= nrFinal:
    print(nrInicial)
    nrInicial += 1
```

### **While com números pares**

<p>Abaixo, temos um código que tem variáveis como "nr" e "max". a var nr recebe o valor 1, enquanto a max recebe um valor ditado pelo usuário. Quando o usuário dita o valor, dará um print "Números pares entre 1 e (max):".
Após, será executado o While.
<i>Enquanto</i> <b>nr</b> for maior ou igual a <b>max</b>:
    recebe um <i>"Se"</i>, que diz:
       <i>Se</i> nr dividido por 2 e o resto for igual a zero:
          Retornará um print de "nr", tendo "end" como a finalização com espaçamento, ao invés de haver uma quebra de linha, como nas células anteriores.
        após o <i>"end"</i>, ele retornará ao loop, assim que o código ler <code> nr += 1 </code>, incrementando mais 1.

Explicando a grosso modo, o número que o usuário digitar será o valor mais 1. Ou seja, se eu digitar 10, o código irá gerar do 1 ao 10, com todos esses valores, ele vai verificar quais números são divisíveis por 2 (dois). Feito isso, ele retornará estes valores, excluindo os que não são divisíveis por dois.
</p>

```python
nr = 1
max = int(input("Digite um número maior que 1: "))

print(f"Números pares entre 1 e {max}:")

while nr <= max:
    #o símbolo "%", neste caso, significa "resto da divisão".
    if nr % 2 == 0:
        print(nr, end=" ")
    nr += 1
```

<p>No caso abaixo, fizemos o contrário, retornando os números ímpares,</P>

```python
nr = 1
max = int(input("Digite um número maior que 1: "))

print(f"Números ímpares entre 1 e {max}:")

while nr <= max:
    #o símbolo "%", neste caso, significa "resto da divisão".
    if nr % 3 == 0:
        print(nr, end=" ")
    nr += 1
```

### **While com Random**

```python
import random
secretnum = random.randint(1,100)
tentativas = 0

while True:
    palpite = int(input("Digite o seu palpite: "))
    tentativas += 1

    if palpite < secretnum:
        print("O número secreto é maior. Tente novamente!")
    elif palpite > secretnum:
        print("O número secreto é menor. Tente novamente!")
    else:
        print(f"Parabéns! O número {palpite} está correto! Você acertou em {tentativas} tentativas")
        break
```

### **Exercício 2 - While com números inteiros**

```python
maiornum = -1  # Começa com -1 para o caso de o usuário digitar 0 logo de início

while True:
    num = int(input("Digite um número inteiro positivo (ou um negativo para sair): "))
    
    if num < 0:
        break
        
    if num > maiornum:
        #Se o num for maior q o maiornum, ele trocará, fazendo com que o maiornum receberá o valor de num digitado no input.
        maiornum = num

if maiornum != -1:
    print(f"O maior número digitado foi: {maiornum}")
else:
    print("Nenhum número positivo foi digitado.") 
```

### **Exercício 3 - Caixa eletrônico**


```python
saldo = 1000000.00

while True:
    print("==========================")
    print("CAIXA ELETRÔNICO | ZBank")
    print("1 - Verificar Saldo")
    print("2 - Depositar dinheiro")
    print("3 - Sacar dinheiro")
    print("4 - Sair")
    opcao = int(input("Escolha uma opção (1-4):"))

# 1 - VERIFICAR SALDO
    if opcao == 1:
            print("___________________________")
            print("VERIFICAR SALDO | ZBank")
            print(f"Seu Saldo é de R$ {saldo: .2f}")
            continue

# 2 - DEPOSITAR DINHEIRO
    elif opcao == 2:
         print("___________________________")
         print("DEPÓSITO DE DINHEIRO | ZBank")
         val_depositado = int(input("Qual valor você deseja depositar? Digite: "))
         if val_depositado > 0:
              saldo += val_depositado
              print(f"Feito! O valor total é: R$ {saldo: .2f}.")
              continue
         else:
              print("Você não pode depositar esse valor! tente novamente")
              continue

# 3 - SACAR DINHEIRO
    elif opcao == 3:
         print("___________________________")
         print("SAQUE DE DINHEIRO | ZBank")
         val_saque = int(input("Digite um valor para saque: "))
         if val_saque > saldo:
              print("Valor excede o saldo disponível na conta! Tente outro valor!")
              continue
         
         elif val_saque <= 0:
              print("Valor solicitado inválido!")
              continue
         else:
              saldo -= val_saque
              print(f"Você conseguiu sacar R$ {val_saque: .2f}!")

# 4 - SAIR
    if opcao == 4:
         print("Tem certeza que deseja sair? (S/N)")
         opcaosaida = str(input("'S' para Sim | 'N' para Não:"))
         if opcaosaida == "S":
              print("Até mais!")
              break
         else:
              continue
    else:
         print("Opção inválida!")
         print("Tente novamente!")
```

# **WHILE COM CONDIÇÕES COMPLEXAS**

O exemplo abaixo demonstra um programa que continua incrementando x e decrementando y
#enquanto x for menor que 10 e y for maior que 10.

Inicializando duas variáveis, x e y
x, y = 5, 15

O loop continuará enquanto ambas as condições forem verdadeiras: x < 10 e y > 10
while x < 10 and y > 10:
    
    print(f"x: {x}, y: {y}") # Imprimindo os valores atuais de x e y
    x += 1 # Incrementando x em 1
    y -= 1 # Decrementando y em 1


```python
x, y = 5, 15

while x < 10 and y > 10:
    print(f"x: {x}, y: {y}")

    #aqui seria a mesma coisa que x = x + 1
    #da mesma forma como y = y - 1
    x += 1
    y -= 1

print("Loop concluído!")
print(f"Valores finais: x = {x}, y = {y}.")
```

```python
import random

secretnum1 = random.randint(1,10)
secretnum2 = random.randint(1,10)

tentativas = 5

adiv1 = False
adiv2 = False

while tentativas > 0 and (not adiv1 or not adiv2):
    print(f"Tentativas restantes: {tentativas}")
    palpite1 = int(input("Adivinhe o primeiro número secreto (1-10): "))
    palpite2 = int(input("Adivinhe o segundo número secreto (1-10): "))
    
    if palpite1 == secretnum1:
        print("Você adivinhou o primeiro número!")
        adiv1 = True

    if palpite2 == secretnum2:
        print("Você adivinhou o segundo número!")
        adiv2 = True

    if not adiv1 or not adiv2:
        print("Tente novamente!")
        tentativas -= 1
if adiv1 == True and adiv2 == True:
    print("Parabéns! Você adivinhou ambos os números.")

else:
    print(f"Você não conseguiu adivinhar ambos os números. Eles eram {secretnum1} e {secretnum2}")
```
