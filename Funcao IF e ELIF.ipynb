```python
#EXERCÍCIO 2

#Idade mímina para votar:
idvote = 18
validade = int(input("Informe sua idade:"))

# "==" é quando o valor deve ser igual
# ">=" ou "<=" é quando o valor de e ser maior ou menor ou igual
if validade >= idvote:
    print("Você tem idade mínima para votar")
else:
    print("Você não tem idade mínima para votar")
```

```python
#Exercício 1

#Verificação de nota de aluno

#Exercício executado com sucesso✅

#Input
nota = int(input("Insira a nota do aluno para verificação: "))

if nota >= 90 and nota <= 100:
    print("Excelente!")
elif nota >= 70 and nota <= 89:
    print("Bom")
elif nota >= 50 and nota <= 69:
    print("Satisfatório")
else:
    print("Nota insulficiente")
```
